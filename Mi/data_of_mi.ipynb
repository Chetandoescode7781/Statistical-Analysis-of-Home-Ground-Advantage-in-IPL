```python
import pandas as pd
import zipfile
import os
```

```python
matches_df = pd.read_csv('../ipl_data/ipl_ball_by_ball_2008_2026.csv')
```

```python
mumbai_team = "Mumbai Indians"

opponent_teams = [
    "Kolkata Knight Riders",
    "Royal Challengers Bengaluru",
    "Royal Challengers Bangalore",
    "Chennai Super Kings"
]

home_grounds = [
    "Wankhede",
    "Eden Gardens",
    "Chinnaswamy",
    "Chidambaram"
]

match_records = []
```

```python
for match_id, group in matches_df.groupby("match_id", sort=False):
    teams = pd.unique(group[["batting_team", "bowling_team"]].to_numpy().ravel())
    teams = [team for team in teams if pd.notna(team)]

    if len(teams) < 2:
        continue

    team1, team2 = teams[:2]
    team_runs = group.groupby("batting_team")["total_runs"].sum()
    team1_runs = int(team_runs.get(team1, 0))
    team2_runs = int(team_runs.get(team2, 0))

    if team1_runs > team2_runs:
        winner = team1
    elif team2_runs > team1_runs:
        winner = team2
    else:
        winner = "Tie"

    match_records.append({
        "match_id": match_id,
        "season": group["season"].iloc[0],
        "date": group["date"].iloc[0],
        "venue": group["venue"].iloc[0],
        "team1": team1,
        "team2": team2,
        "team1_runs": team1_runs,
        "team2_runs": team2_runs,
        "winner": winner
    })

match_summary_df = pd.DataFrame(match_records)
```

```python
is_mumbai_match = (
    ((match_summary_df["team1"] == mumbai_team) & (match_summary_df["team2"].isin(opponent_teams))) |
    ((match_summary_df["team2"] == mumbai_team) & (match_summary_df["team1"].isin(opponent_teams)))
)

filtered_df = match_summary_df[is_mumbai_match].copy()

venue_pattern = "|".join(home_grounds)
filtered_df = filtered_df[
    filtered_df["venue"].str.contains(venue_pattern, case=False, na=False)
]
```

```python
extracted_data = []

for _, row in filtered_df.iterrows():
    if row["team1"] == mumbai_team:
        runs = row.get("team1_runs", 0)
    else:
        runs = row.get("team2_runs", 0)

    if row["winner"] == mumbai_team:
        win_loss = "Win"
    elif pd.isna(row["winner"]) or row["winner"] == "Tie":
        win_loss = "Tie/No Result"
    else:
        win_loss = "Loss"

    is_home = "Wankhede" in str(row["venue"])

    extracted_data.append({
        "team": row["team2"] if row["team1"] == mumbai_team else row["team1"],
        "ground": row["venue"],
        "runs": runs,
        "win/loss": win_loss,
        "is_home": is_home
    })
```

```python
result_df = pd.DataFrame(extracted_data)
display(result_df)
```
